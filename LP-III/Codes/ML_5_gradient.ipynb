```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3)**2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2
learning_rate = 0.1
iterations = 50

for i in range(iterations):
    grad = derivative(x)
    x = x - learning_rate * grad

    print("Iteration:", i + 1, "x:", round(x, 6),
          "y:", round(function(x), 6))
```

```python
print("\nLocal Minimum:")
print("x =", round(x, 6))
print("y =", round(function(x), 6))
```

```python
x_values = np.linspace(-6, 4, 100)
y_values = function(x_values)

plt.plot(x_values, y_values, label='y = (x+3)^2')
plt.scatter(x, function(x), color='red', label='Minimum')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent')
plt.legend()
plt.grid()
plt.show()
```

```python
# Gradient Descent is an optimization algorithm used to find the minimum value of a function.
# It is commonly used in Machine Learning to reduce errors and improve model performance.
# It starts with an initial value and repeatedly updates it using the derivative of the function.
# The derivative indicates the direction and slope of the function.
# Formula: New Value = Current Value - (Learning Rate * Derivative)
# Learning Rate controls the size of each update step.
# Derivative indicates the direction in which the function increases or decreases.
# Iterations represent the number of times the value is updated.
# Working:
# 1. Select an initial value for x.
# 2. Calculate the derivative of the function.
# 3. Multiply the derivative by the learning rate.
# 4. Subtract the result from the current value to get the new value.
# 5. Repeat the process until the minimum point is reached.
# In this practical, the function is y = (x + 3)^2 and the initial value is x = 2.
# Gradient Descent gradually moves x towards -3, where the minimum value of y is 0.
```
